```python
from sklearn.datasets import load_digits
import pandas as pd
dataset = load_digits()
dataset.keys()
```

```python
df = pd.DataFrame(dataset.data)
df.head()
```

```python
x= df
y=dataset.target
```

```python
y
```

```python
from sklearn.preprocessing import StandardScaler
scaler = StandardScaler()
x_scaled = scaler.fit_transform(x)
x_scaled
```

```python
from sklearn.linear_model import LogisticRegression
from sklearn.model_selection import train_test_split

x_train, x_test, y_train, y_test = train_test_split(x_scaled, y, test_size=0.2, random_state=42)

model = LogisticRegression(max_iter=1000)
model.fit(x_train, y_train)
model.score(x_test, y_test)
```

```python
from sklearn.decomposition import PCA

pca = PCA(0.95)
x_pca = pca.fit_transform(x)
x_pca.shape
```

```python
pca.explained_variance_ratio_
```

```python
pca.n_components_
```

```python
X_train_pca, X_test_pca, y_train, y_test = train_test_split(x_pca, y, test_size=0.2, random_state=30)
```

```python
model = LogisticRegression(max_iter=1000)
model.fit(X_train_pca, y_train)
model.score(X_test_pca, y_test)
```

LDA : Linear Discrimination Analysis



```python
from sklearn.discriminant_analysis import LinearDiscriminantAnalysis as lda
from sklearn.model_selection import train_test_split
from sklearn.metrics import accuracy_score
from sklearn.datasets import load_iris
import time

```

```python
dataset = load_iris()
dataset.keys()
```

```python
df = pd.DataFrame(dataset.data)
df.head()
```

```python
df.describe()
```

```python
df.shape
```

```python
dataset.data[0].reshape(2,2)
```

```python
x = df
y = dataset.target
```

```python
lda_model = LinearDiscriminantAnalysis(n_components = 1)
```

```python
start_time = time.time()
```

```python
X_train, X_test, Y_train, Y_test = train_test_split(x, y, test_size=0.2, random_state=42)
X_train_lda = lda_model.fit_transform(X_train, Y_train)
X_test_lda = lda_model.transform(X_test)
```

```python
lda_model = LogisticRegression()
lda_model.fit(X_train_lda, Y_train)
```

```python
lda_pred = lda_model.predict(X_test_lda)
```

```python
lda_time = time.time()- start_time
```

```python
predictions_for_accuracy = lda_model.predict(X_test_lda)
lda_accuracy = accuracy_score(Y_test, predictions_for_accuracy)
```

```python
print("LDA Accuracy: ", round(lda_accuracy, 4))
print("LDA Time: ", round(lda_time, 2), "seconds")
```
